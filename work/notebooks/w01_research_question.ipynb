## 3. Quick look at the data (2-3 real numbers)

```python
import os, sys, subprocess
from rich import print


IN_COLAB = "google.colab" in sys.modules
REPO_URL = "https://github.com/flyrank-bih/flyrank-ml-internship-starter"
REPO_DIR = "flyrank-ml-internship-starter"


if IN_COLAB:
   if not os.path.isdir(REPO_DIR):
       subprocess.run(["git", "clone", "--depth", "1", REPO_URL, REPO_DIR], check=True)
   os.chdir(REPO_DIR)
   subprocess.run([sys.executable, "-m", "pip", "install", "-q", "-r", "requirements.txt"], check=True)
elif os.path.basename(os.getcwd()) == "notebooks":
   os.chdir("..")


import pandas as pd, numpy as np
df = pd.read_csv("data/raw/content_refresh_anonymized.csv")

print("Preference of long content")
df['long_content']= df['word_count']>=2000
new_df=df.groupby('long_content')['impressions_90d'].mean()
print(new_df.to_string())
print("\n\nThis shows that pages which have a words higher than a certain number have almost double impressions compared to the ones that don't!!")

#skewness of sites
print("\n\nSkewness of sites")
df_sorted = df.sort_values(by='impressions_90d', ascending=False).reset_index(drop=True)

total_clicks = df_sorted['impressions_90d'].sum()
df_sorted['cumulative_impressions'] = df_sorted['impressions_90d'].cumsum()
df_sorted['cumulative_percent'] = df_sorted['cumulative_impressions'] / total_clicks

top_performers = df_sorted[df_sorted['cumulative_percent'] <= 0.80]
percent_of_urls = (len(top_performers) / len(df_sorted)) * 100

print(f"Just {percent_of_urls:.1f}% of the URLs are driving 80% of all organic impressions.")


```
